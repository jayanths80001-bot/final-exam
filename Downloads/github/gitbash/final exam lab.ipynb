```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
text="AI and ML pathway is for 'CS' student!. AI have a Machine Learning"
token=word_tokenize(text)
print("tokenization:",token)
```

```python
#stopword Removelabs
filter=[w for w in token if w.isalpha() and w.lower() not in stopwords.words("english")]
print("after removal Stopwords:",filter)
```

```python
#punctuation removel
import string
punct_remove=[w for w in token if w not in string.punctuation]
print("after remove punctuation",punct_remove)
```

```python
#pos_tag
from nltk import pos_tag
text="karnataka have a red and yellow plag"
token=word_tokenize(text)

filter=[w for w in token if w.isalpha() and w.lower() not in stopwords.words("english")]
print("pos_tag:",pos_tag(filter))
```

```python
#NER
import spacy
text="karnataka have a red and yellow plag and have a 30% forest"
token=word_tokenize(text)

nlp=spacy.load("en_core_web_sm")
doc=nlp(text)

print("Name entity recognition:",[(ent.text,ent.label_)for ent in doc.ents])
```

```python
#LEMMATIZATION
from nltk.stem import WordNetLemmatizer
WNT=WordNetLemmatizer()
words=["playing","plays","students","programming"]
for word in words:
    print("Lemmatization:",WNT.lemmatize(word,pos='v'))
```
